# EDA — crypto-market-elt

Exploración de los marts generados por el pipeline. Requiere haber corrido `make pipeline` antes
(el warehouse vive en `warehouse/crypto.duckdb`, fuera de Git).

```python
import duckdb

conn = duckdb.connect("../warehouse/crypto.duckdb", read_only=True)
conn.sql("SHOW ALL TABLES").df()
```

```python
# Retornos y volatilidad por símbolo
ohlcv = conn.sql("SELECT * FROM main_marts.mart_daily_ohlcv ORDER BY symbol, trade_date").df()
ohlcv.groupby("symbol")[["daily_return", "volatility_30d"]].describe()
```

```python
# Precio de cierre por símbolo
ohlcv.pivot_table(index="trade_date", columns="symbol", values="close").plot(
    subplots=True, figsize=(10, 12), title="Precio de cierre (USDT)"
);
```
